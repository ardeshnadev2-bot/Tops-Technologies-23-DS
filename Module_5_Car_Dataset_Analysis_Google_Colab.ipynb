# Step 1: Import Required Libraries


```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import LabelEncoder
```


# Step 2: Load Dataset

```python

# Replace filename if needed
df = pd.read_csv('data.csv')

df.head()
```


# Step 3: Basic Information About Dataset


```python

print("Shape of Dataset:", df.shape)

df.info()

df.describe()
```


# Step 4: Check Missing Values


```python

df.isnull().sum()
```


# Step 5: Handle Missing Values


```python

# Fill numerical missing values with median
num_cols = df.select_dtypes(include=np.number).columns

for col in num_cols:
    df[col].fillna(df[col].median(), inplace=True)

# Fill categorical missing values with mode
cat_cols = df.select_dtypes(include='object').columns

for col in cat_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)

df.isnull().sum()
```


# Step 6: Remove Duplicate Rows


```python

print("Duplicate Rows:", df.duplicated().sum())

df.drop_duplicates(inplace=True)

print("Dataset Shape After Removing Duplicates:", df.shape)
```


# Step 7: Data Visualization


```python

plt.figure(figsize=(10,6))

sns.histplot(df['MSRP'], bins=50)

plt.title('Distribution of Car Prices (MSRP)')
plt.xlabel('MSRP')
plt.ylabel('Frequency')

plt.show()
```

```python

plt.figure(figsize=(12,6))

top_brands = df['Make'].value_counts().head(10)

sns.barplot(x=top_brands.index, y=top_brands.values)

plt.title('Top 10 Car Brands')
plt.xticks(rotation=45)

plt.show()
```

```python

plt.figure(figsize=(12,8))

corr = df.select_dtypes(include=np.number).corr()

sns.heatmap(corr, annot=True, cmap='coolwarm')

plt.title('Correlation Heatmap')

plt.show()
```


# Step 8: Outlier Detection


```python

plt.figure(figsize=(10,5))

sns.boxplot(x=df['MSRP'])

plt.title('Outlier Detection for MSRP')

plt.show()
```


# Step 9: Encode Categorical Features


```python

label_encoder = LabelEncoder()

for col in cat_cols:
    df[col] = label_encoder.fit_transform(df[col])

df.head()
```


# Step 10: Feature Selection


```python

X = df.drop('MSRP', axis=1)

y = df['MSRP']

print("Features Shape:", X.shape)
print("Target Shape:", y.shape)
```


# Step 11: Final Cleaned Dataset


```python

df.head()

# Save cleaned dataset
df.to_csv('cleaned_car_dataset.csv', index=False)

print("Cleaned dataset saved successfully.")
```
